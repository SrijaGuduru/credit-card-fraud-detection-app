Sequence Creation

```python
X_train_df = pd.DataFrame(X_train)
X_test_df = pd.DataFrame(X_test)

y_train = y_train.reset_index(drop=True)
y_test = y_test.reset_index(drop=True)
```

```python
import numpy as np

def create_sequences(X, y, time_steps=10):
    X_seq = []
    y_seq = []

    for i in range(len(X) - time_steps):
        X_seq.append(X.iloc[i:i+time_steps].values)
        y_seq.append(y[i+time_steps])

    return np.array(X_seq), np.array(y_seq)
```

```python
time_steps = 10

X_train_seq, y_train_seq = create_sequences(X_train_df, y_train, time_steps)
X_test_seq, y_test_seq = create_sequences(X_test_df, y_test, time_steps)
```

```python
print("Train shape:", X_train_seq.shape)
print("Test shape:", X_test_seq.shape)
```

TRAIN LSTM MODEL:Import Required Libraries

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.metrics import Precision, Recall
```

Build LSTM Model

```python
time_steps = X_train_seq.shape[1]
num_features = X_train_seq.shape[2]

model = Sequential()

model.add(LSTM(64, return_sequences=True,
               input_shape=(time_steps, num_features)))
model.add(Dropout(0.3))

model.add(LSTM(32))
model.add(Dropout(0.3))

model.add(Dense(1, activation='sigmoid'))

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=[Precision(), Recall()]
)

model.summary()
```

Train Model

```python
history = model.fit(
    X_train_seq, y_train_seq,
    epochs=5,              # start with 5 (faster)
    batch_size=128,
    validation_data=(X_test_seq, y_test_seq)
)
```

Evaluate Model

```python
loss, precision, recall = model.evaluate(X_test_seq, y_test_seq)

print("Precision:", precision)
print("Recall:", recall)
```

Generate Confusion Matrix

```python
from sklearn.metrics import confusion_matrix
import numpy as np

y_pred = model.predict(X_test_seq)
y_pred = (y_pred > 0.5).astype(int)

cm = confusion_matrix(y_test_seq, y_pred)
print(cm)
```

Import PyTorch

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

Convert Data to PyTorch Tensors

```python
X_train_seq
y_train_seq
```

```python
X_train_tensor = torch.tensor(X_train_seq, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train_seq, dtype=torch.float32)

X_test_tensor = torch.tensor(X_test_seq, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test_seq, dtype=torch.float32)
```

Create Liquid Layer

```python
class LiquidLayer(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(LiquidLayer, self).__init__()
        self.W = nn.Linear(hidden_size, hidden_size)
        self.U = nn.Linear(input_size, hidden_size)

    def forward(self, x, h):
        dh = -h + torch.tanh(self.W(h) + self.U(x))
        h = h + dh
        return h
```

Create Liquid Neural Network Model

```python
class LiquidNet(nn.Module):
    def __init__(self, input_size, hidden_size):
        super(LiquidNet, self).__init__()
        self.hidden_size = hidden_size
        self.liquid = LiquidLayer(input_size, hidden_size)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        batch_size = x.size(0)
        h = torch.zeros(batch_size, self.hidden_size)

        for t in range(x.size(1)):
            h = self.liquid(x[:, t, :], h)

        out = torch.sigmoid(self.fc(h))
        return out
```

Initialize Model

```python
input_size = X_train_seq.shape[2]
hidden_size = 64

liquid_model = LiquidNet(input_size, hidden_size)

criterion = nn.BCELoss()
optimizer = optim.Adam(liquid_model.parameters(), lr=0.001)
```

Train Liquid Model

```python
epochs = 5

for epoch in range(epochs):
    optimizer.zero_grad()

    outputs = liquid_model(X_train_tensor)
    loss = criterion(outputs.squeeze(), y_train_tensor)

    loss.backward()
    optimizer.step()

    print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item()}")
```

Evaluate Liquid Model

```python
with torch.no_grad():
    outputs = liquid_model(X_test_tensor)
    predictions = (outputs.squeeze() > 0.5).float()

accuracy = (predictions == y_test_tensor).sum().item() / len(y_test_tensor)
print("Accuracy:", accuracy)
```

Combine Predictions (Hybrid Score)

Final Decision Threshold

Evaluate Hybrid Model

```python
from sklearn.metrics import confusion_matrix, classification_report

print("Confusion Matrix:")
print(confusion_matrix(y_test_seq, hybrid_pred))

print("\nClassification Report:")
print(classification_report(y_test_seq, hybrid_pred))
```
